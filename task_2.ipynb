# Task 2 - Recursive Utility Functions
I denne oppgaven lager vi to rekursive funksjoner. 
Den første funksjonen finner loggfiler i mapper og undermapper. 
Den andre funksjonen undersøker JSON-data og finner aktive kasseregistre. 

Rekursjon betyr at en funksjon kaller på seg selv for å løse en oppgave. 

```python
from pathlib import Path
import json
```

## Task 2.1 - Recursive Directory Search

Her skal vi lage en funsjon som finner alle filtypene med "log" i en mappe og dens undermapper.
Vi bruker Path fra pathlib og rekursjon for å undersøke mappestrukturen.
Funksjonen skal returnere en lise med Path-objekter. 

```python
def find_log_files(folder: Path) -> list[Path]: 
    """
    Finner alle log filene i en mappe og undermappene.
    Args: (folder: Path) : Mappen som skal undersøkes.
    Returns: list[Path]: en liste med filbaner tl loggfilene.
    """
    
    folder = Path(folder)
    log_files = []

# Går igjennom alle elementene i mappen
    for item in folder.iterdir():

        # Hvis elemtet er en mappe, undersøkes den rekursivt
        if item.is_dir():
            log_files.extend(find_log_files(item))

        # Hvis elemtet er en .log fil, legges den til i listen
        elif item.is_file() and item.suffix == ".log":
            log_files.append(item)

    return log_files
```

```python
log_files = find_log_files(Path("data"))

print("Antall loggfiler:", len(log_files))

for file in log_files:
    print(file)
```

### Resultat
Funksjonen find_log_files() fant totalt 9 loggfiler i datamappen. Resultatet viser at funskjonen kan søke gjennom undermapper ved hjelp av rekurasjon og finne filer med fil-endingen .log. 

## Task 2.2 - Recursive JSON Traversal
Lager en funskjon som undersøker JSON-data og finner ID ene til alle aktive kasseregistre. 

JSON-filen inneholder både dictionaries og lister på flere nivåer, derfor 
bruker vi rekursjon for å undersøke hele datastrukturen. 

Resultatet skal returneres som et set, sånn at vi unngår dublikanter. 

```python
# Leser JSON filen med informasjon om butikker og kasseregistre 
with open("data/store_config.json", "r", encoding="utf-8") as file:
    config = json.load(file)
```

```python
def find_active_registers(data: object) -> set[str]:

    """
    Finner ID ene til aktive kassregistre i en datastruktur.
    Funksjonen undersøker dictionares og lister rekursivt.
    Args: (data: object): Datene som skal undersøkes.
    Returns: set[str]: Et sett med id`er til aktive kasseregistre.
    """

    active_registers = set()

    # Hvis data er en dictionary undersøkes innholdet
    if isinstance(data, dict):

        # Legger til ID`en dersom kasseregisteret er aktivt
        if data.get("active") is True and "id" in data:
            active_registers.add(data["id"])

        # Undersøker alle verdiene i dictionaryen rekursivt
        for value in data.values():
            active_registers.update(find_active_registers(value))

    # Hvis data er en liste undersøkes hhvert element
    elif isinstance(data, list):
        for item in data:
            active_registers.update(find_active_registers(item))

    return active_registers
```

```python
active_registers = find_active_registers(config)

print("Aktive kasseregistre:", sorted(active_registers))
print("Antall aktive kasseregistre:", len(active_registers))
```

### Resultat
Funksjonen find_active_registers() fant tre aktive kasseregistre i store_config.json: 'POS_01', 'POS_02', 'POS_03'.

Resultatet viser at funksjonen klarer å undersøke både dictionaries og lister ved hjelp av rekursjon og hente ut ID`ene til aktive kasseregistre.
