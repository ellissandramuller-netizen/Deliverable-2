## Task 3.1 - Outlier Detection and Correction

I denne oppgaven bruker vi NumPy til å finne priser som avviker betydelig fra gjennomsnittet.

Vi definerer en outliner som en pris som ligger mer enn tre standaravvik fra gjennomsnittet. 
Disse prisene skal erstattes med medianen av de gyldige prisene. 

```python
import numpy as np

def correct_outliers(prices: np.ndarray) -> np.ndarray:
    
    """
    Finner priser som avviker mer enn tre standardavvik fra gjennomsnittet
    og erstatter dem med madianen av de gyldige prisene.

    Args: (prices: np.ndarray): En array med priser

    Returns: np.ndarray: En array med korrigerte priser.
    """
    prices = np.array(prices, dtype=float, copy=True)

    # Bergener gjennomsnittet og standardavvik
    mean = np.mean(prices)
    std = np.std(prices)

    # Finner priser som ligger mer enn tre standardavvik fra gjennomsnittet 
    outliers = np.abs(prices - mean) > (3 * std)

    # Finner prisene som ikke er outliers
    valid_prices = prices[~outliers]

    # Bergener medianen av de gyldige prisene
    median = np.median(valid_prices)

    # Erstatter outliers med medianen
    prices[outliers] = median

    return prices
```

```python
# Lager flere vanlige priser og en unormalt høy pris
test_prices = np.array([30] * 20 + [500], dtype=float)

# Korrigerer prisene
resultat = correct_outliers(test_prices)

print("Opprinnelig siste pris:", test_prices[-1])
print("Korrigert siste pris:", resultat[-1])
```
